```python
import pandas as pd

# upload students_marks.csv first (Files panel > Upload)
df = pd.read_csv("students_marks.csv")
print(df.head())
print(df.shape)

subjects = ["Math", "Stats", "Python", "Databases", "English"]
averages = df[subjects].mean()
print(averages)
print("Highest average:", averages.idxmax())
print("Below 50 in Math:", (df["Math"] < 50).sum())
```

## Task 3

```python
import matplotlib.pyplot as plt

plt.bar(subjects, averages)
plt.title("Average marks by subject")
plt.xlabel("Subject")
plt.ylabel("Average marks")
plt.ylim(0, 100)
plt.show()

import pandas, matplotlib
print("pandas", pandas.__version__)
print("matplotlib", matplotlib.__version__)
```
