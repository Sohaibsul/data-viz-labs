## Task 2

```python
subjects = ["Math", "Stats", "Python", "Databases", "English"]
marks = [72, 85, 90, 78, 66]
print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
print("Highest:", max(marks))
print("Lowest:", min(marks))
print("Top subject:", subjects[marks.index(max(marks))])
for s, m in zip(subjects, marks):
    print(s, "Pass" if m >= 50 else "Fail")
```

## Task 3

```python
import matplotlib.pyplot as plt
plt.bar(subjects, marks, color="steelblue")
plt.title("Marks by subject")
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.show()
```

## Task 4

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
hours = [2, 3, 1, 4, 3, 5, 2]
plt.plot(days, hours, marker="o")
plt.title("Study hours this week")
plt.xlabel("Day")
plt.ylabel("Hours")
plt.show()
print("Total hours:", sum(hours))
print("Most hours on:", days[hours.index(max(hours))])
```
